## Basics of Python

### Basic data types

#### Numbers

Integers and floats work as you would expect from other languages:

```python
# Create a variable and check what type Python gave it.
x = 3
print(x, type(x))
```

```python
# The basic arithmetic operators.
print(x + 1)   # Addition
print(x - 1)   # Subtraction
print(x * 2)   # Multiplication
print(x ** 2)  # Exponentiation
```

```python
# Shorthand for updating a variable in place: x += 1 means x = x + 1.
x += 1
print(x)
x *= 2
print(x)
```

```python
# Floats (decimal numbers) work the same way.
y = 2.5
print(type(y))
print(y, y + 1, y * 2, y ** 2)
```

Note that unlike many languages, Python does not have unary increment (x++) or decrement (x--) operators.

Python also has built-in types for long integers and complex numbers; you can find all of the details in the [documentation](https://docs.python.org/3.7/library/stdtypes.html#numeric-types-int-float-long-complex).

#### Booleans

Python implements all of the usual operators for Boolean logic, but uses English words rather than symbols (`&&`, `||`, etc.):

```python
# Booleans hold one of two values: True or False.
t, f = True, False
print(type(t))
```

Now we let's look at the operations:

```python
# Python spells the logical operators as English words.
print(t and f) # Logical AND;
print(t or f)  # Logical OR;
print(not t)   # Logical NOT;
print(t != f)  # Logical XOR;
```

#### Strings

```python
# Create strings and measure how long they are.
hello = 'hello'   # String literals can use single quotes
world = "world"   # or double quotes; it does not matter
print(hello, len(hello))
```

```python
# Join strings together with +.
hw = hello + ' ' + world  # String concatenation
print(hw)
```

```python
# Insert values into a string with .format() - you will see this in lab output.
hw12 = '{} {} {}'.format(hello, world, 12)  # string formatting
print(hw12)
```

String objects have a bunch of useful methods; for example:

```python
# String methods you will actually reach for when cleaning text data.
s = "hello"
print(s.capitalize())  # Capitalize a string
print(s.upper())       # Convert a string to uppercase; prints "HELLO"
print(s.replace('l', '(ell)'))  # Replace all instances of one substring with another
print('  world '.strip())  # Strip leading and trailing whitespace
```

You can find a list of all string methods in the [documentation](https://docs.python.org/3.7/library/stdtypes.html#string-methods).

### Containers

Python includes several built-in container types: lists, dictionaries, sets, and tuples.

#### Lists

A list is the Python equivalent of an array, but is resizeable and can contain elements of different types:

```python
# Create a list, then read items out of it by position.
xs = [3, 1, 2]   # Create a list
print(xs, xs[2])
print(xs[-1])     # Negative indices count from the end of the list; prints "2"
```

```python
# A list can hold mixed types, and any item can be replaced.
xs[2] = 'foo'    # Lists can contain elements of different types
print(xs)
```

```python
# Add an item to the end of a list.
xs.append('bar') # Add a new element to the end of the list
print(xs)
```

```python
# Remove the last item and hand it back.
x = xs.pop()     # Remove and return the last element of the list
print(x, xs)
```

As usual, you can find all the gory details about lists in the [documentation](https://docs.python.org/3.7/tutorial/datastructures.html#more-on-lists).

#### Slicing

In addition to accessing list elements one at a time, Python provides concise syntax to access sublists; this is known as slicing:

```python
# Slicing pulls a range of items out of a list.
nums = list(range(5))    # range is a built-in function that creates a list of integers
print(nums)         # Prints "[0, 1, 2, 3, 4]"
print(nums[2:4])    # Get a slice from index 2 to 4 (exclusive); prints "[2, 3]"
print(nums[2:])     # Get a slice from index 2 to the end; prints "[2, 3, 4]"
print(nums[:2])     # Get a slice from the start to index 2 (exclusive); prints "[0, 1]"
print(nums[:])      # Get a slice of the whole list; prints ["0, 1, 2, 3, 4]"
print(nums[:-1])    # Slice indices can be negative; prints ["0, 1, 2, 3]"
nums[2:4] = [8, 9]  # Assign a new sublist to a slice
print(nums)         # Prints "[0, 1, 8, 9, 4]"
```

#### Loops

You can loop over the elements of a list like this:

```python
# Loop over the items of a list one at a time.
animals = ['cat', 'dog', 'monkey']
for animal in animals:
    print(animal)
```

If you want access to the index of each element within the body of a loop, use the built-in `enumerate` function:

```python
# Use enumerate() when you also need each item's position.
animals = ['cat', 'dog', 'monkey']
for idx, animal in enumerate(animals):
    print('#{}: {}'.format(idx + 1, animal))
```

#### List comprehensions

When programming, frequently we want to transform one type of data into another. As a simple example, consider the following code that computes square numbers:

```python
# Build a list of squares the long way, with a loop.
nums = [0, 1, 2, 3, 4]
squares = []
for x in nums:
    squares.append(x ** 2)
print(squares)
```

You can make this code simpler using a list comprehension:

```python
# The same result as a list comprehension - one readable line.
nums = [0, 1, 2, 3, 4]
squares = [x ** 2 for x in nums]
print(squares)
```

List comprehensions can also contain conditions:

```python
# A list comprehension can filter as it goes.
nums = [0, 1, 2, 3, 4]
even_squares = [x ** 2 for x in nums if x % 2 == 0]
print(even_squares)
```

#### Dictionaries

A dictionary stores (key, value) pairs, similar to a `Map` in Java or an object in Javascript. You can use it like this:

```python
# A dictionary stores key -> value pairs and looks values up by key.
d = {'cat': 'cute', 'dog': 'furry'}  # Create a new dictionary with some data
print(d['cat'])       # Get an entry from a dictionary; prints "cute"
print('cat' in d)     # Check if a dictionary has a given key; prints "True"
```

```python
# Add a new key to a dictionary.
d['fish'] = 'wet'    # Set an entry in a dictionary
print(d['fish'])      # Prints "wet"
```

```python
# Asking for a key that does not exist raises a KeyError - this cell fails on purpose.
print(d['monkey'])  # KeyError: 'monkey' not a key of d
```

```python
# Use .get() to supply a fallback instead of failing when a key is missing.
print(d.get('monkey', 'N/A'))  # Get an element with a default; prints "N/A"
print(d.get('fish', 'N/A'))    # Get an element with a default; prints "wet"
```

```python
# Delete a key from a dictionary.
del d['fish']        # Remove an element from a dictionary
print(d.get('fish', 'N/A')) # "fish" is no longer a key; prints "N/A"
```

You can find all you need to know about dictionaries in the [documentation](https://docs.python.org/2/library/stdtypes.html#dict).

It is easy to iterate over the keys in a dictionary:

```python
# Loop over a dictionary's keys and values together.
d = {'person': 2, 'cat': 4, 'spider': 8}
for animal, legs in d.items():
    print('A {} has {} legs'.format(animal, legs))
```

Dictionary comprehensions: These are similar to list comprehensions, but allow you to easily construct dictionaries. For example:

```python
# A dictionary comprehension - the dictionary version of a list comprehension.
nums = [0, 1, 2, 3, 4]
even_num_to_square = {x: x ** 2 for x in nums if x % 2 == 0}
print(even_num_to_square)
```

### Functions

Python functions are defined using the `def` keyword. For example:

The two examples here are deliberately small. You will see `def` used on real data in the **Pandas** section below.

```python
# Define a function with def, then call it on several inputs.
def sign(x):
    if x > 0:
        return 'positive'
    elif x < 0:
        return 'negative'
    else:
        return 'zero'

for x in [-1, 0, 1]:
    print(sign(x))
```

We will often define functions to take optional keyword arguments, like this:

```python
# A function argument can carry a default that the caller may override.
def hello(name, loud=False):
    if loud:
        print('HELLO, {}'.format(name.upper()))
    else:
        print('Hello, {}!'.format(name))

hello('Bob')
hello('Fred', loud=True)
```

## Numpy

Numpy is the core library for numerical computing in Python. It provides a fast multidimensional array object and tools for working with these arrays. Pandas, which we cover next, is built on top of it.

To use Numpy, we first need to import the `numpy` package:

```python
# Import numpy. 'as np' is the universal shorthand - every lab uses it.
import numpy as np
```

### Arrays

A numpy array is a grid of values, all of the same type, and is indexed by a tuple of nonnegative integers. The number of dimensions is the rank of the array; the shape of an array is a tuple of integers giving the size of the array along each dimension.

We can initialize numpy arrays from nested Python lists, and access elements using square brackets:

```python
# Create a one-dimensional array and inspect its shape and its items.
a = np.array([1, 2, 3])  # Create a rank 1 array
print(type(a), a.shape, a[0], a[1], a[2])
a[0] = 5                 # Change an element of the array
print(a)
```

```python
# A two-dimensional array is a list of rows.
b = np.array([[1,2,3],[4,5,6]])   # Create a rank 2 array
print(b)
```

```python
# .shape reports (rows, columns); index into it with [row, column].
print(b.shape)
print(b[0, 0], b[0, 1], b[1, 0])
```

Numpy also provides many functions to create arrays:

```python
# Create an array pre-filled with zeros.
a = np.zeros((2,2))  # Create an array of all zeros
print(a)
```

```python
# The same, filled with ones.
b = np.ones((1,2))   # Create an array of all ones
print(b)
```

```python
# The same, filled with a constant you choose.
c = np.full((2,2), 7) # Create a constant array
print(c)
```

```python
# The identity matrix - ones down the diagonal.
d = np.eye(2)        # Create a 2x2 identity matrix
print(d)
```

```python
# Random values between 0 and 1; your numbers will differ every run.
e = np.random.random((2,2)) # Create an array filled with random values
print(e)
```

### Array indexing

Numpy offers several ways to index into arrays.

Boolean array indexing: Boolean array indexing lets you pick out arbitrary elements of an array. Frequently this type of indexing is used to select the elements of an array that satisfy some condition. Here is an example:

```python
# Comparing an array to a value gives a True/False array of the same shape.
a = np.array([[1,2], [3, 4], [5, 6]])

bool_idx = (a > 2)
print(bool_idx)
```

```python
# Use that True/False array to pull out only the values that matched.
print(a[bool_idx])
print(a[a > 2])
print(bool_idx)
```

### Array math

Basic mathematical functions operate elementwise on arrays, and are available both as operator overloads and as functions in the numpy module:

```python
# Arithmetic on arrays happens element by element.
x = np.array([[1,2],[3,4]], dtype=np.float64)
y = np.array([[5,6],[7,8]], dtype=np.float64)

# Elementwise sum; both produce the array
print(x)
print(y)
print(x + y)
print(np.add(x, y))
```

```python
# Elementwise difference - the operator and the numpy function do the same thing.
print(x - y)
print(np.subtract(x, y))
```

```python
# Elementwise product. Note this is NOT matrix multiplication.
print(x * y)
print(np.multiply(x, y))
```

```python
# Elementwise division.
print(x / y)
print(np.divide(x, y))
```

```python
# Elementwise square root.
print(np.sqrt(x))
```

Numpy provides many useful functions for performing computations on arrays; one of the most useful is `sum`:

```python
# Sum everything, or sum down the columns (axis=0) or across the rows (axis=1).
x = np.array([[1,2],[3,4]])
print(x)
print(np.sum(x))  # Compute sum of all elements; prints "10"
print(np.sum(x, axis=0))  # Compute sum of each column; prints "[4 6]"
print(np.sum(x, axis=1))  # Compute sum of each row; prints "[3 7]"
```

## Pandas

Numpy arrays hold numbers. Real business data almost never arrives that way — it arrives as a **table**: named columns, mixed types, one row per customer, order, or store.

**Pandas** is the library for those tables, and its `DataFrame` is the most important object in this course. Nearly every lab you will do this term starts by loading one. Think of a DataFrame as a spreadsheet you drive with code: named columns, numbered rows.

```python
# Import pandas. 'as pd' is the universal shorthand, just like np and plt.
import pandas as pd
```

```python
# Build a small DataFrame from a dictionary: each key becomes a column.
sales = pd.DataFrame({
    'region':  ['East', 'East', 'West', 'West', 'North', 'North'],
    'product': ['Widget', 'Gadget', 'Widget', 'Gadget', 'Widget', 'Gadget'],
    'units':   [120, 85, 210, 45, 60, 150],
    'price':   [9.99, 24.50, 9.99, 24.50, 9.99, 24.50],
})
sales
```

That is the same dictionary idea you met earlier: the keys became the column names and the lists became the columns.

```python
# The first three things to check about any new table: its size, its columns, its first rows.
print(sales.shape)                 # (rows, columns)
print(sales.columns.tolist())
sales.head(3)
```

```python
# Select a single column by name, or several by passing a list of names.
print(sales['units'])
sales[['product', 'units']]
```

```python
# Add a new column computed from existing ones - the arithmetic runs row by row.
sales['revenue'] = sales['units'] * sales['price']
sales
```

```python
# Keep only the rows where a condition is True.
sales[sales['revenue'] > 2000]
```

```python
# .describe() reports the standard summary statistics for every numeric column at once.
sales.describe()
```

```python
# Group the rows by one column, then total another column within each group.
print(sales.groupby('region')['units'].sum())
print()
print(sales.groupby('region')['revenue'].sum())
```

### Writing your own function

You met `def` earlier on toy examples. Here is the reason to bother with it: as soon as you catch yourself writing the same line twice with one word changed, that line wants to be a function.

```python
# Wrap the grouping in a function so the same summary can be run on any column.
def revenue_by(table, column):
    return table.groupby(column)['revenue'].sum()

print(revenue_by(sales, 'region'))
print()
print(revenue_by(sales, 'product'))
```

### Plotting

The most important function in `matplotlib` is plot, which allows you to plot 2D data. Here is a simple example:

```python
# Plot a sine curve: build the x values, compute y, then draw.
x = np.arange(0, 3 * np.pi, 0.1)
y = np.sin(x)

plt.plot(x, y)
```

With just a little bit of extra work we can easily plot multiple lines at once, and add a title, legend, and axis labels:

```python
# Plot two lines on the same axes, then label them so a reader can tell them apart.
y_sin = np.sin(x)
y_cos = np.cos(x)

# Plot the points using matplotlib
plt.plot(x, y_sin)
plt.plot(x, y_cos)
plt.xlabel('x axis label')
plt.ylabel('y axis label')
plt.title('Sine and Cosine')
plt.legend(['Sine', 'Cosine'])
```

### Subplots

You can plot different things in the same figure using the subplot function. Here is an example:

```python
# Compute the x and y coordinates for points on sine and cosine curves
x = np.arange(0, 3 * np.pi, 0.1)
y_sin = np.sin(x)
y_cos = np.cos(x)

# Set up a subplot grid that has height 2 and width 1,
# and set the first such subplot as active.
plt.subplot(1, 2, 1)

# Make the first plot
plt.plot(x, y_sin)
plt.title('Sine')

# Set the second subplot as active, and make the second plot.
plt.subplot(1, 2, 2)
plt.plot(x, y_cos)
plt.title('Cosine')

# Show the figure.
plt.show()
```
